###Initializing Library

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
print("Library Initialized!")
```

### Synthesizing Data

```python
np.random.seed(358)
n_samples = 50
X1 = np.random.rand(n_samples) * 2
X2 = np.random.rand(n_samples) * 3
noise = np.random.rand(n_samples) * 1.5

#Linear Influence
Y = 1.79 + 3.58 * X1 - 2 * X2 + noise

#Fusing into DataFrame
df = pd.DataFrame({'X1': X1, 'X2': X2, 'Y': Y})
display(df.head())
```

###Fitting Model

```python
X = df[['X1', 'X2']]
y = df['Y']

#Splitting training 80% and testing 20%
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=358)

#Fitting model
model = LinearRegression()
model.fit(X_train, y_train)

#Printing intercept and coefficient
print(f"Intercept (Constant): {model.intercept_:.4f}")
print(f"Coefficients X1: {model.coef_[0]:.4f}")
print(f"Coefficients X2: {model.coef_[1]:.4f}")

#Y = 2.46 + 3.57 X1 - 1.96 X2
```

###Matrix and Prediction

```python
#Prediction
y_pred = model.predict(X_test)

#Matrix
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

#Print Matrix
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"R-squared (R2): {r2:.4f}")
```

###Visualization

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(x=y_test, y=y_pred, color='blue', alpha=0.7)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], color = "red", linestyle="--", lw=2)
plt.xlabel('Actual Values (Y)')
plt.ylabel('Predicted Values (Y_pred)')
plt.title('Actual vs. Predicted Values')
plt.grid(True)
plt.show()
```
